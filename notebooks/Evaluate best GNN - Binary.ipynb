```python
import torch
from argparse import Namespace
from src.main import train_gnn
import itertools
import pandas as pd
from argparse import Namespace
from tqdm import tqdm
```

```python
gnn_dims = [64, 256, 1024]
gnn_num_layers = [2, 3, 4]
gnn_use_bn = [True, False]
gnn_dropout = [0, 0.1]
gnn_aggr = ["sum", "mean"]
lr = [0.001, 0.0001]
gnn_model = ['sage', 'gat', 'gatv2', 'transformer']

# Hyperparameter combinations
param_grid = itertools.product(
    gnn_dims,
    gnn_num_layers,
    #gnn_use_bn,
    #gnn_dropout,
    gnn_aggr,
    lr,
    #gnn_model,
)

# Base args Namespace
base_args = Namespace(
    dataset_path="/Users/sebastian/University/Master/third term/sem-proj/kg-token/data/ml-100k/",
    gnn_hidden_dim=128,
    gnn_out_dim=128,
    gnn_num_layers=2,
    gnn_dropout=0.0,
    gnn_num_heads=1,
    gnn_use_bn=False,
    aggr='sum',
    lr=0.0001,
    wd=0.05,
    seed=0,
    num_epochs=250,
    gnn_model="gat",
    device="cpu",
)

results = []

for params in tqdm(param_grid, desc="Hyperparameter Tuning", unit="config"):
    args = base_args
    args.gnn_hidden_dim = params[0]
    args.gnn_out_dim = params[0]
    args.gnn_num_layers = params[1]
    args.gnn_use_bn = False
    args.gnn_dropout = 0
    args.aggr = params[2]
    args.lr = params[3]
    args.gnn_model = 'gat'
    
    train_acc, test_acc, test_loss = train_gnn(args)
    
    results.append({
        "gnn_hidden_dim": args.gnn_hidden_dim,
        "gnn_out_dim": args.gnn_out_dim,
        "gnn_num_layers": args.gnn_num_layers,
        "gnn_use_bn": args.gnn_use_bn,
        "gnn_dropout": args.gnn_dropout,
        "gnn_aggr": args.aggr,
        "lr": args.lr,
        "gnn_model": args.gnn_model,
        "train_accuracy": train_acc,
        "test_accuracy": test_acc,
        "test_loss": test_loss,
    })

results_df = pd.DataFrame(results)

print(results_df)
```

```python
results_df.to_csv('gat.csv', index=False)
```
