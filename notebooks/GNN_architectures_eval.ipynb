```python
import torch
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Binary (Likes/Doesnt Predictions)

```python
df = pd.read_csv('/Users/sebastian/University/Master/third term/sem-proj/kg-token/src/results/results_binary.csv', index_col=0)
```

```python
df.sort_values(by="test_accuracy", ascending=False).head(5)
```

```python
plt.figure(figsize=(10, 6))
models = df['gnn_model'].unique()
for model in models:
    subset = df[df['gnn_model'] == model]
    plt.scatter(subset['gnn_hidden_dim'], subset['test_accuracy'], label=model, alpha=0.7)
plt.xlabel('GNN Hidden Dimension')
plt.ylabel('Test Accuracy')
plt.title('Test Accuracy vs. Hidden Dimension')
plt.legend(title='GNN Model')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
models = df['gnn_model'].unique()
for model in models:
    subset = df[df['gnn_model'] == model]
    plt.scatter(subset['gnn_num_layers'], subset['test_accuracy'], label=model, alpha=0.7)
plt.xlabel('GNN Number of Layers')
plt.ylabel('Test Accuracy')
plt.title('Test Accuracy by GNN Model and Number of Layers')
plt.legend(title='GNN Model')
plt.grid(True)
plt.show()
```
